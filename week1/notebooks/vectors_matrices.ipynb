```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
p1 = np.array([1, 2, 3])
p2 = np.array([4, 5, 6])

v = p2 - p1
print(v)
```

### visualization

```python
fig = plt.figure()
ax = fig.add_subplot(projection="3d")

ax.quiver(p1[0], p1[1], p1[2], v[0], v[1], v[2], color="b", label="Vector v")

ax.scatter(p1[0], p1[1], p1[2], color="red", s=50, label="p1 (Start)")
ax.scatter(p2[0], p2[1], p2[2], color="green", s=50, label="p2 (End)")

ax.set_xlim([0, 6])
ax.set_ylim([0, 6])
ax.set_zlim([0, 6])

ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")
ax.set_zlabel("Z Axis")

ax.legend()
plt.show()
```
