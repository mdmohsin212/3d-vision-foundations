```python
import numpy as np
```

```python
A = np.array([
    [1.0, 2.0, 1.0],
    [3.0, 8.0, 1.0],
    [0.0, 4.0, 1.0]
])
print(A)
```

```python
E21 = np.eye(3)

multiplier = A[1, 0] / A[0, 0]
```

```python
E21[1, 0] = -multiplier

print("Constructed Elementary Matrix E21:\n", E21)
```

```python
A_after_E21 = E21 @ A

print("Result of E21 @ A :\n", A_after_E21)
```
