```python
import numpy as np
```

```python
A = np.array([
    [2.0, 1.0],
    [4.0, 5.0]
])

print(A)
```

```python
pivot_1 = A[0, 0]
entry_to_eliminate = A[1, 0]
```

```python
multiplier = entry_to_eliminate / pivot_1

print(f"Pivot 1: {pivot_1}")
print(f"Multiplier: {multiplier}")
```

```python
A[1] = A[1] - multiplier * A[0]

print("Matrix A after elimination (Row 2 first entry is now 0):", A)
```
