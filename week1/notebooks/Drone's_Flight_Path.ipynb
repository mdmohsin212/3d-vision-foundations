```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
pointA = np.array([2, 3, 8])
pointB = np.array([7, 9, 2])
```

```python
flight = pointB - pointA
print(flight)
```

```python
fig = plt.figure()
ax = fig.add_subplot(projection="3d")

ax.quiver(pointA[0], pointA[1], pointA[2], flight[0], flight[1], flight[2], color="orange", label="Flight Vector")

ax.scatter(pointA[0], pointA[1], pointA[2], s=50, color="red", label="(Start Point A)")
ax.scatter(pointB[0], pointB[1], pointB[2], s=50, color="green", label="(End Point B)")

ax.set_xlim([0, 10])
ax.set_ylim([0, 10])
ax.set_zlim([0, 10])

ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")
ax.set_zlabel("Z Axis")

fig.legend()
plt.show()
```
