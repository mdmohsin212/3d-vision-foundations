```python
import numpy as np
```

```python
def dot_product(a, b):
    return np.sum(a * b)

def vector_norm(a):
    return np.sqrt(np.sum(a * a))
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(dot_product(a, b))
print("\n", vector_norm(a))
```

```python
print(np.dot(a, b))
print(np.linalg.norm(a))
```


`cos(theta) = (a · b) / (||a|| ||b||)`

```python
def angle_between(a, b):
    cos_theta = dot_product(a, b) / (vector_norm(a) * vector_norm(b))
    cos_theta = np.clip(cos_theta, -1.0, 1.0)
    theta = np.arccos(cos_theta)
    return np.degrees(theta)
```

```python
print("Angle between : ", angle_between(a, b))
```

*Cross product + distance*

```python
def cross_product(a, b):
    return np.cross(a, b)

def distance_between_points(p1, p2):
    return vector_norm(p2 - p1)
```

```python
print("Cross Product: ", cross_product(a, b))
print("Distance between points: ", distance_between_points(a, b))
```
